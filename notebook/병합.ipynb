```
import pandas as pd

crime = pd.read_csv("crime_2023.csv", encoding="cp949")
drink = pd.read_excel("drinking_2023.xlsx")
high = pd.read_excel("drinking_highrisk_2023.xlsx")
pop = pd.read_excel("population.xlsx")

crime_melt = crime.melt(
    id_vars=["범죄대분류", "범죄중분류"],
    var_name="지역",
    value_name="범죄건수"
)

crime_melt["범죄건수"] = crime_melt["범죄건수"].fillna(0)

drink = drink.rename(columns={
    "시군구별": "지역",
    "응답자수": "음주_응답자수",
    "조율": "음주율(%)"
})

high = high.rename(columns={
    "시군구별": "지역",
    "응답자수": "고위험_응답자수",
    "조율": "고위험음주율(%)"
})

pop = pop.rename(columns={
    "행정구역별": "지역",
    "계 (명)": "인구수"
})

pop["인구수"] = pop["인구수"].astype(str).str.replace(",", "").astype(int)

df = crime_melt.merge(pop[["지역", "인구수"]], on="지역", how="left")
df = df.merge(drink[["지역", "음주율(%)", "음주_응답자수"]], on="지역", how="left")
df = df.merge(high[["지역", "고위험음주율(%)", "고위험_응답자수"]], on="지역", how="left")

df = df[
    [
        "지역",
        "인구수",

        "음주_응답자수",
        "음주율(%)",
        "고위험_응답자수",
        "고위험음주율(%)",

        "범죄대분류",
        "범죄중분류",
        "범죄건수",
    ]
]

df.to_csv("final_merged_dataset.csv", index=False, encoding="utf-8-sig")
```

```
import pandas as pd

crime = pd.read_csv("crime_2023.csv", encoding="cp949")
drink = pd.read_excel("drinking_2023.xlsx")
high = pd.read_excel("drinking_highrisk_2023.xlsx")
pop = pd.read_excel("population.xlsx")

crime_melt = crime.melt(
    id_vars=["범죄대분류", "범죄중분류"],
    var_name="지역",
    value_name="범죄건수"
)

crime_melt["범죄건수"] = crime_melt["범죄건수"].fillna(0)

drink = drink.rename(columns={
    "시군구별": "지역",
    "응답자수": "음주_응답자수",
    "조율": "음주율(%)"
})

high = high.rename(columns={
    "시군구별": "지역",
    "응답자수": "고위험_응답자수",
    "조율": "고위험음주율(%)"
})

pop = pop.rename(columns={
    "행정구역별": "지역",
    "계 (명)": "인구수"
})

pop["인구수"] = pop["인구수"].astype(str).str.replace(",", "").astype(int)

crime_sum = crime_melt.groupby("지역", as_index=False)["범죄건수"].sum()
crime_sum = crime_sum.rename(columns={"범죄건수": "총범죄건수"})

df = pop.merge(drink[["지역", "음주_응답자수", "음주율(%)"]], on="지역", how="left")
df = df.merge(high[["지역", "고위험_응답자수", "고위험음주율(%)"]], on="지역", how="left")
df = df.merge(crime_sum, on="지역", how="left")

df = df[
    [
        "지역",
        "인구수",
        "음주_응답자수",
        "음주율(%)",
        "고위험_응답자수",
        "고위험음주율(%)",
        "총범죄건수",
    ]
]

df.to_csv("final_clean_dataset.csv", index=False, encoding="utf-8-sig")
```
